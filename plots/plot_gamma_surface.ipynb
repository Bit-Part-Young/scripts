### GSFE γ-surface 绘制

数据来源：https://www.ctcms.nist.gov/potentials/atomman/tutorial/4.5._Gamma_surface_plotting.html

30x30 grid GSFE data

```python
import pandas as pd
import numpy as np

df = pd.read_csv("gamma_surface.csv")
df.shape
```

```python
ncols, nrows = 30, 30
X = df.iloc[:, 0].values.reshape(ncols, nrows)
Y = df.iloc[:, 1].values.reshape(ncols, nrows)
Z = df.iloc[:, 2].values.reshape(ncols, nrows)
# eV/Angstrom^2 -> mJ/m^2
Z = Z * 1.6021766208 * 10**4
```

```python
import matplotlib.pyplot as plt
from spt.plot_params import set_plot_params

set_plot_params(roman_params=True, sci_params=True)
fig, ax = plt.subplots()

# SFE 范围和分段
levels = np.linspace(0, int(Z.max()), 10)

# 填充等高线图
contourf = ax.contourf(X, Y, Z, levels=levels, cmap=plt.get_cmap("jet"))

# 等高线（可选）
contour = ax.contour(X, Y, Z, levels=levels, colors="k", linewidths=0.5, alpha=0.8)

# colorbar
cbar = fig.colorbar(contourf, ax=ax, ticks=levels, label="SFE (mJ/m$^2$)")

ax.set_xlabel(r"1/2[$\bar{1}$10]")
ax.set_ylabel(r"1/2[11$\bar{2}$]")
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.grid(True, linestyle="--", alpha=0.5, which="both")
# ax.set_aspect("equal", adjustable="box")
```
